```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Load the CSV file
file_path = 'two_loops_sma_cable_3jan2025.csv'
data = pd.read_csv(file_path)

# Convert 'Time (s)' to a usable datetime format
data['Time (s)'] = pd.to_datetime(data['Time (s)'], format='%H,%M,%S')

# Select 10 equally spaced iterations
unique_iterations = data['Iteration'].unique()
selected_iterations = np.linspace(0, len(unique_iterations) - 1, 5, dtype=int)
selected_iterations = unique_iterations[selected_iterations]

# Prepare for plotting
fig, axes = plt.subplots(2, 1, figsize=(10, 12), sharex=True)

# Use a colormap for different colors
colors = cm.viridis(np.linspace(0, 1, len(selected_iterations)))

# Plot 20*log10(magnitude of S11) and 20*log10(magnitude of S21) for selected iterations
for idx, iteration in enumerate(selected_iterations):
    iter_subset = data[data['Iteration'] == iteration]
    frequency = iter_subset['Frequency (Hz)']
    
    # Calculate magnitude and then 20*log10 for S11 and S21
    s11_magnitude = np.sqrt(iter_subset['S11 Real']**2 + iter_subset['S11 Imag']**2)
    s21_magnitude = np.sqrt(iter_subset['S21 Real']**2 + iter_subset['S21 Imag']**2)
    
    s11_log = 20 * np.log10(s11_magnitude)
    s21_log = 20 * np.log10(s21_magnitude)
    
    # Get the first timestamp for the iteration
    timestamp = iter_subset['Time (s)'].iloc[0].strftime('%H:%M:%S')
    
    # Plot with a specific color from the colormap
    axes[0].plot(frequency, s11_log, label=f'Time {timestamp}', color=colors[idx])
    axes[1].plot(frequency, s21_log, label=f'Time {timestamp}', color=colors[idx])

# Set labels and titles
axes[0].set_title('20*log10(|S11|)')
axes[0].set_ylabel('Magnitude (dB)')
axes[0].legend(loc='lower right', fontsize='small')

axes[1].set_title('20*log10(|S21|)')
axes[1].set_xlabel('Frequency (Hz)')
axes[1].set_ylabel('Magnitude (dB)')
axes[1].legend(loc='upper right', fontsize='small')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Load the CSV file
file_path = 'two_loops_sma_cable_3jan2025.csv'
data = pd.read_csv(file_path)

# Convert 'Time (s)' to a usable datetime format
data['Time (s)'] = pd.to_datetime(data['Time (s)'], format='%H,%M,%S')

# Select 10 equally spaced iterations
unique_iterations = data['Iteration'].unique()
selected_iterations = np.linspace(0, len(unique_iterations) - 1, 5, dtype=int)
selected_iterations = unique_iterations[selected_iterations]

# Example temperature array corresponding to the iterations (replace with actual values)
#temperatures = [23.5, 24.0, 24.2, 24.5, 25.0, 25.3, 25.7, 26.0, 26.2, 26.5]

# Prepare for plotting
fig, axes = plt.subplots(2, 1, figsize=(10, 12), sharex=True)

# Use a colormap for different colors
colors = cm.viridis(np.linspace(0, 1, len(selected_iterations)))

# Store legend information
custom_legends_s11 = []
custom_legends_s21 = []

# Plot 20*log10(magnitude of S11) and 20*log10(magnitude of S21) for selected iterations
for idx, iteration in enumerate(selected_iterations):
    iter_subset = data[data['Iteration'] == iteration]
    frequency = iter_subset['Frequency (Hz)']
    
    # Calculate magnitude and then 20*log10 for S11 and S21
    s11_magnitude = np.sqrt(iter_subset['S11 Real']**2 + iter_subset['S11 Imag']**2)
    s21_magnitude = np.sqrt(iter_subset['S21 Real']**2 + iter_subset['S21 Imag']**2)
    
    s11_log = 20 * np.log10(s11_magnitude)
    s21_log = 20 * np.log10(s21_magnitude)
    
    # Get the corresponding temperature for the iteration
    temperature = temperatures[idx]
    
    # Plot with a specific color from the colormap
    line_s11, = axes[0].plot(frequency, s11_log, color=colors[idx])
    line_s21, = axes[1].plot(frequency, s21_log, color=colors[idx])
    
    # Append to custom legend with temperature
    custom_legends_s11.append((line_s11, f'Temp {temperature} K'))
    custom_legends_s21.append((line_s21, f'Temp {temperature} K'))

# Add custom legends
axes[0].legend(*zip(*custom_legends_s11), loc='lower right', fontsize='large')
axes[1].legend(*zip(*custom_legends_s21), loc='upper right', fontsize='large')

# Set labels and titles
axes[0].set_title('Return loss: |S11|',fontweight='bold')
axes[0].set_ylabel('Magnitude (dB)',fontweight='bold')

axes[1].set_title('Insertion loss: |S21|',fontweight='bold')
axes[1].set_xlabel('Frequency (Hz)',fontweight='bold')
axes[1].set_ylabel('Magnitude (dB)',fontweight='bold')

for ax in axes:
    ax.tick_params(axis='both', which='major', labelsize=10)
    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_fontweight('bold')

plt.tight_layout()
plt.show()
```
