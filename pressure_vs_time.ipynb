```python
# if logged just pressure
counter=0
pressure_readings=np.array([])
date_time=np.array([])
with open(filename+'.xls', mode ='r')as file:
  csvFile = csv.reader(file)
  for lines in csvFile:
      counter+=1
      if(counter>6):
          lines1=lines[0].split('\t')
          #print(lines1)
          pressure_readings=np.append(pressure_readings,10**3*float(lines1[0]))
          date_time=np.append(date_time,lines1[1])
```

```python
data_time1=np.empty(len(date_time),dtype='U10')
for i in range(len(date_time)):
    data_time1[i]=date_time[i].split(' ')[1]
```

```python
start_p=4000
stop_p=-50
fig,ax=plt.subplots(figsize=(10,6))
plt.plot(data_time1[start_p:stop_p],pressure_readings[start_p:stop_p],'b-',linewidth=2)
ax.xaxis.set_major_locator(MaxNLocator(nbins=12))
plt.xticks(rotation=70)
#plt.title("Cryo-cooler temperature curve",fontsize='large', fontweight='bold')#- 13/09/24")
plt.xlabel("Date-time",fontsize='16', fontweight='bold')
plt.ylabel("Pressure near pump in mbar",fontsize='16', fontweight='bold')
plt.xticks(fontsize = 14, fontweight='bold')
plt.yticks(fontsize = 14, fontweight='bold')
plt.grid()
plt.show()
```
